# 문제 5. 복합 조건
각각 조건을 만족하는 행 수를 출력하세요.

1. 3등석 **그리고** 남성 **그리고** 생존한 승객 (`&`)
2. 1등석 **또는** 요금 100 이상인 승객 (`|`)
3. 혼자 탑승하지 **않은** 승객 (`~`, `alone` 열 사용)

```python
cond_01 = (titanic["pclass"] == 3) & (titanic["sex"] == "male") & (titanic["survived"] == 1 )
print(cond_01.sum())
print(cond_01)
cond_02 = (titanic["pclass"] == 1) | (titanic["fare"] >= 100)
print(cond_02.sum())
print(cond_02)
cond_03 = ~titanic["alone"]
print(cond_03.sum())
print(cond_03)
```

# 문제 6. isin - between
1. `isin`으로 Cherbourg 또는 Queenstown에서 탑승한 승객을 골라 `shape`를 출력하세요.
2. `between`으로 나이 20~30세(양 끝 포함) 승객을 골라 `shape`와 나이의 min·max를 출력하세요.
3. 같은 범위를 양 끝 미포함(`inclusive="neither"`)으로 바꿔 `shape`를 출력하세요.

```python
mask_isin = titanic["embark_town"].isin(["Cherbourg", "Queenstown"])
result = titanic[mask_isin]
print(result.shape)

mask_between = titanic["age"].between(20, 30)
result_02 = titanic[mask_between]

print("20~30세 승객 : ", result_02.shape)
print("나이 최솟값 : ", result_02["age"].min())
print("나이 최대값 : ", result_02["age"].max())

mask_between_neither = titanic["age"].between(20, 30, inclusive="neither")
result_03 = titanic[mask_between_neither]
print("20, 30 미포함 승객 : ", result_03.shape)
```

# 문제 7. loc[조건, 열]
12살 이하 그리고 3등석 승객의 age, sex, survived 열만 kids 변수에 담고, 행 수와 생존률(survived 평균, 소수 둘째 자리)을 출력하세요.

```python
cond_kid_mask = (titanic["age"] <= 12) & (titanic["pclass"] == 3)
kids = titanic.loc[cond_kid_mask, [ "age" , "sex" , "survived" ]]

print("행 수 : ", len(kids))

survival_kid = kids["survived"].mean()
print(f"생존률 : {survival_kid:.2f}")
```

# 문제 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석
질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

1. 남성 · 나이 20~40세(`between`) · 1·2등석(`isin`) 조건으로 `pclass`, `age`, `fare`, `survived` 열을 선택하세요.
2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
3. 같은 조건을 `query`로 써서 행 수를 검증하세요.
4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.

```python
cond_mask = (titanic["sex"] == "male") & (titanic["age"].between(20, 40)) & (titanic["pclass"].isin([1, 2]))
result_03 = titanic.loc[cond_mask, ["pclass", "age", "fare", "survived"]]

# 행수
print("행 수:", len(result_03))

# 생존률
survival_rate = result_03["survived"].mean()
print(f"생존률: {survival_rate:.2f}")
print(result_03.sort_values(by="fare", ascending=False).head(3))

# query로 문구로 변형
query_03 = titanic.query("sex == 'male' and 20 <= age <= 40 and pclass in [1, 2]")
print("query 행 수:", len(query_03))
```

```python
# 3등석 조건 생존률
cond_04 = (titanic["sex"] == "male") & (titanic["age"].between(20, 40)) & (titanic["pclass"].isin([3]))
result_04 = titanic.loc[cond_04, ["survived"]]

survival_04 = result_04["survived"].mean()
print(f"3등석 남성(20~40세) 생존률: {survival_04:.2f}")
```

# 비교해석 (1·2등석 vs 3등석)
1,2등석 남성과 3등석 남성의 생존률의 차이는 9%가 차이가 난다. 각 등급 별로 객실이 위치하는 곳이 탈출 할 수 있는 경로와 관련이 있어 보여지며, 생존에 있어 큰 영향력이 있어보입니다.
